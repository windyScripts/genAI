list comprehension: Like map.
[x**2 for x in range(10)]

basic syntax:

[expression for item in iterable]

with conditional logic: [expression for item in iterable if condition]


````python
square = [num**2 for num in range(10) if num > 4]
print(square)
````

nested list comprehension

[expression for item1 in iterable1 for item2 in iterable2]
Two lists, comprehension.

````python
## Nested list comprehension

lst1 = [1,2,3,4]
lst2 = ['a','b','c','d']

pair = [(i,j) for i in lst1 for j in lst2]
print(pair) # Prints all possible pairs.
````

## Dictionaries

Intro, creation, accessing, modifying, methods, iterating, nested, comprehensions

unordered collections of items, key value pairs, keys must be unique and immutable. Values can be of any type.

curvy braces for dictionaries

dictionaries are mutable.

 dictionaries are shared by address with =.

 shallow copy: 


````python
empty_dict = {}
empty_dict2 = dict()
print(type(empty_dict))

student = {"name":"wind","age":50,"grade":"Titanium","address":"India"}
print(student)

student1 = {"name":"wind","age":50,"name":"Titanium"}
print(student1)

print(student['age'])
print(student.get('grade'))
print(student.get('last_name','not Available'))

del student['age'] # deletes age key value
keys = student.keys()
values = student.values()
items = student.items()
print(keys, values, items)

# shallow copy
student_copy = student.copy() # creates a separate copy in memory.
for key, value in student.items():
    print(f"{key}:{value}")
````

File operations:

````python
with open('example.txt','r') as file:
    content = file.read()
    print(content)
````

````python
# read file line by line:
with open('example.txt','r') as file:
    i = 0
    for line in file:
        i+=1
        print(line.strip(), i) # removes newlines
````

````python
## overwriting a file

with open('example.txt','w') as file:
    file.write('Hello World \n')
    file.write('This is a new line. \n')
````

````python
# write without overwriting

with open('example.txt','a') as file:
    file.write("Append operation taking place\n")

with open('example.txt','r') as file:
        for line in file:
            print(line.strip())

````

````python
# writing lines to a file
lines = ['first line \n', 'second line \n']
with open('example.txt','a') as file:
    for line in lines:
        file.write(line)
````

````python
# writing lines to a file
lines = ['first line \n', 'second line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
````

````python
# Binary files

data = b'\x00\x01\x02i\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
````

````python
# read file and write to a destination file:
with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
````

The w+ mode is used to open a file for both reading and writing. If the file doesn't exist, it will be created, and if it does exist, its content is truncated.

````python
with open('example2.txt','w+') as file:
    file.write("Hello world\n")
    file.write("Tjis is a new line \n")
    # move file cursor to beginning.
    file.seek(0)

    content = file.read()
    print(content)
````
